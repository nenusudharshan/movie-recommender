```python
import pandas as pd

movies = pd.read_csv("datasets/tmdb_5000_movies.csv")
movies.head()
```

```python
credits = pd.read_csv("datasets/tmdb_5000_credits.csv")
credits.head()
```

```python
movies.shape
```

```python
credits.shape
```

```python
movies = movies.merge(credits,on='title')
movies.info()
```

```python
movies = movies[['movie_id','title','overview','genres','keywords','cast','crew']]
```

```python
movies.info()
```

```python
movies.isnull().sum()
```

```python
movies.dropna(inplace=True)
```

```python
movies.duplicated().sum()
```

```python
import ast
def convert(obj):
    if isinstance(obj, list):   
        return obj
    
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
```

```python
movies['genres'] = movies['genres'].apply(convert)
movies.head()
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
movies.head()
```

```python
def convert_cast(obj):
    if isinstance(obj, list):
        return obj
    L=[]
    for i in ast.literal_eval(obj)[:3]:
        L.append(i['name'])
    return L
```

```python
movies['cast'] = movies['cast'].apply(convert_cast)
movies.head()
```

```python
movies['cast'].head(2)
```

```python
def get_director(obj):
    if isinstance(obj, list):
        return obj
    L = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            L.append(i['name'])
            break
    return L        
```

```python
movies['crew'] = movies['crew'].apply(get_director)
```

```python
movies['crew'].head(3)
```

```python
movies.head()
```

```python
movies['overview'] = movies['overview'].apply(lambda x: x.split())
```

```python
movies.head(1)
```

```python
def clean_list(L):
    return [i.replace(" ","") for i in L]

movies['genres'] = movies['genres'].apply(clean_list)
movies['cast'] = movies['cast'].apply(clean_list)
movies['keywords'] = movies['keywords'].apply(clean_list)
movies['crew'] = movies['crew'].apply(clean_list)
```

```python
movies.head()
```

```python
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['cast']
```

```python
movies.head()
```

```python
new_movies_df = movies[['movie_id','title','tags']].copy()
```

```python
new_movies_df.head()
```

```python
new_movies_df['tags'] = new_movies_df['tags'].apply(lambda x: " ".join(x).lower() if isinstance(x,list) else x)
```

```python
new_movies_df.head(3)
```

```python
new_movies_df
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem_words(text):
    return " ".join([ps.stem(word) for word in text.split()])
```

```python
new_movies_df['tags'] = new_movies_df['tags'].apply(stem_words)
```

```python
# 1) Re-vectorize using the final stemmed tags
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000, stop_words='english')
vectors = cv.fit_transform(new_movies_df['tags']).toarray()   # NOTE: use 'vectors' variable name
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vectors)
```

```python
similarity[119]
```

```python
from difflib import get_close_matches

title_to_index = pd.Series(new_movies_df.index, index=new_movies_df['title']).to_dict()

def recommend(movie_title, top_n=5):
    if movie_title not in title_to_index:
        print(f"Movie '{movie_title}' not found.")
        matches = get_close_matches(movie_title, new_movies_df['title'].tolist(), n=5, cutoff=0.4)
        if matches:
            print("Did you mean:", matches)
        return []

    idx = title_to_index[movie_title]
    distances = similarity[idx]
    movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1: top_n+1]

    results = []
    for movie_idx, score in movies_list:
        title = new_movies_df.iloc[movie_idx]['title']
        results.append(title)

    for title in results:
        print(f"{title}")

```

```python
recommend('Batman Begins')
```
